```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
data = pd.read_excel('C:/Users/lenovo/Desktop/携程民宿信息.xlsx',engine='openpyxl')
```

```python
data['总评分'] = data['总评分'].str.replace('分','')
data['总评论数'] = data['总评论数'].str.replace('条点评','')
data['人数'] = data['人数'].str.replace('人','')
data['床数'] = data['床数'].str.replace('床','')
data.shape
```

```python
data
```

```python
#添加新列
data['超赞房东'] = ''
data['免费停车'] = ''
data['无接触入住'] = ''
data['是否靠近地铁'] = ''
data['是否靠近景区'] = ''
data['评分等级'] = ''
data.shape
```

```python
for i in range(len(data['总评分'])):
    if float(data['总评分'][i])>4.5:
        data['评分等级'][i] = 1
    else:
        data['评分等级'][i] = 0
```

```python
data
```

```python
for i in range(len(data['条件'])):
    if '超赞房东' in str(data['条件'][i]):
        data['超赞房东'][i] = 1
    else:
        data['超赞房东'][i] = 0
```

```python
for i in range(len(data['条件'])):
    if '无接触入住' in str(data['条件'][i]):
        data['无接触入住'][i] = 1
    else:
        data['无接触入住'][i] = 0
```

```python
for i in range(len(data['条件'])):
    if '近地铁' in str(data['条件'][i]):
        data['是否靠近地铁'][i] = 1
    else:
        data['是否靠近地铁'][i] = 0
```

```python
for i in range(len(data['条件'])):
    if '景区' in str(data['条件'][i]):
        data['是否靠近景区'][i] = 1
    else:
        data['是否靠近景区'][i] = 0
```

```python
Newdata = data.dropna(subset=['总评论数','入住便捷','高性价比','位置便利','服务周到','干净卫生','价格'])#处理缺失值
Newdata.shape
```

```python
mydata = Newdata[['入住便捷','位置便利','服务周到','干净卫生','评分等级']]
mydata.head()
```

```python
mydata.columns = ['convenient','location','service','health','total']
mydata
```

```python
#多元回归 分析获得客户评价对推荐人数的影响
import statsmodels.api as sm
from sklearn.linear_model import LinearRegression
x = mydata[['convenient','location','service','health']]
y = mydata[['total']]
regr = LinearRegression()
regr.fit(x,y)
print('各项系数'+str(regr.coef_))
print('常数项'+str(regr.intercept_))
x2 = sm.add_constant(x)
est = sm.OLS(y.astype(float),x2.astype(float)).fit()
print(est.summary())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, roc_curve, confusion_matrix, classification_report
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
x = mydata.drop(columns='total')
y = mydata['total']
x_train,x_test,y_train,y_test = train_test_split(x,y.astype(float),test_size=0.2,random_state=1)
model = DecisionTreeClassifier(max_depth=4,random_state=1)
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
a = pd.DataFrame()
a['预测值'] = list(y_pred)
a['实际值'] = list(y_test)
print(a)
#准确度
score = accuracy_score(y_pred,y_test)
print(score)
#模型评估
fpr,tpr,thres = roc_curve(y_test,y_pred)
a = pd.DataFrame()
a['阈值'] = list(thres)
a['假报警'] = list(fpr)
a['命中率'] = list(tpr)
print(a)
#获取特征重要性
fea = x.columns
importances = model.feature_importances_
#以二维表格显示
importances_df = pd.DataFrame()
importances_df['特征名称'] = fea
importances_df['特征重要性'] = importances
importances_df.sort_values('特征重要性',ascending=False)
print(importances_df)

# 计算混淆矩阵
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Predicted 0', 'Predicted 1'], 
            yticklabels=['Actual 0', 'Actual 1'])
plt.title('Confusion Matrix')
plt.ylabel('Actual Label')
plt.xlabel('Predicted Label')
plt.tight_layout()
plt.show()
```

```python
from sklearn import tree
clf = tree.DecisionTreeClassifier()
clf = clf.fit(x.astype(float),y.astype(float))
clf.score(x.astype(float),y.astype(float))
```

```python
import matplotlib.pyplot as plt
from sklearn import tree

featureName = ['convenient', 'location', 'service', 'health']
className = ['great', 'bad']

# 创建图形
plt.figure(figsize=(12, 8))
tree.plot_tree(clf, 
               feature_names=featureName,
               class_names=className,
               filled=True,
               rounded=True,
               fontsize=10)
plt.title("Decision Tree Visualization")
plt.tight_layout()

# 保存图像
plt.savefig('C:/Users/lenovo/Desktop/dectree12.png', dpi=300, bbox_inches='tight')

# 显示图像
plt.show()
```
